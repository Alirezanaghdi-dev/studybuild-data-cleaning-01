```python
import pandas as pd
```

```python
file_path = "First Dataset.xlsx"
xls = pd.ExcelFile(file_path)
df = pd.read_excel(xls, sheet_name=xls.sheet_names[0])
```

```python
df = df.drop_duplicates(subset=['customer_id'], keep='first')
```

```python
df['age'] = df['age'].fillna(df['age'].median())
```

```python
df['total_spending'] = df['total_spending'].fillna(df['purchase_count'] * df['avg_order_value'])
```

```python
df['signup_date'] = pd.to_datetime(df['signup_date'])
```

```python
text_columns = ['first_name', 'city', 'province', 'membership_tier', 'payment_method', 'device', 'discount_used']
for col in text_columns:
    df[col] = df[col].astype(str).str.strip().str.title()
```

```python
output_path = "Cleaned_First_Dataset.xlsx"
df.to_excel(output_path, index=False)
```
